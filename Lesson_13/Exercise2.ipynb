<p style="font-size:20px;">*** Raster datasets may contain one or more bands. Each band stores a different type of information.</p>

## Create a single band1 object of the rasterio dataset

```python
# Reading the Band Data
band1 = dataset_tiff.read(1)

# Calculate statistics for the first band
band1_stats = {
        'min': band1.min(),
        'mean': band1.mean(),
        'median': np.median(band1),
        'max': band1.max()
    }

# Print the statistics for Band 1
print("Statistics for Band 1:")
print(f"  Min: {band1_stats['min']}")
print(f"  Mean: {band1_stats['mean']}")
print(f"  Median: {band1_stats['median']}")
print(f"  Max: {band1_stats['max']}")
```

#### Basic plotting

```python
# Plot band 1
rasterio.plot.show((band1),cmap='Reds',)
```

```python
# Plot band 2
rasterio.plot.show(dataset_tiff.read(2),cmap='Greens',)
```

```python
# Setting Up the Visualization
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 6))
# Displaying the Band with Different Color Maps
ax1.imshow(band1, cmap='gray')
ax1.set_title('Gray Scale')
ax2.imshow(band1, cmap='viridis')
ax2.set_title('Viridis')
ax3.imshow(band1, cmap='hot')
ax3.set_title('Hot')
```

## Creating and Visualizing Color Composites

#### Create an RGB Color Composite:

RGB composites combine three raster bands into a colour image suitable for visual interpretation.

```python
# Read the red, green, and blue bands
red = dataset_tiff.read(1)
green = dataset_tiff.read(2)
blue = dataset_tiff.read(3)

# Function to normalize the band data
def normalize(arr):
    arr_min = arr.min()
    arr_max = arr.max()
    return (arr - arr_min) / (arr_max - arr_min)

# Stack and normalize the bands for an RGB composite

red=normalize(red)
green=normalize(green)
blue=normalize(blue)

  # Stack bands in the correct order
rgb_normalized = np.dstack((red,green,blue))


# Display the image
plt.imshow(rgb_normalized)
plt.title('RGB Color Composite')
plt.show()
```
