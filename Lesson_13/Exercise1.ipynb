```python
# Import required libraries
import geopandas as gpd
import matplotlib.pyplot as plt
import contextily as ctx
```

### STEP    1: Reading the Shapefile

```python
# Read file using gpd.read_file()
data = gpd.read_file("./Input_Data/shapeFile/Clipping_polygon.shp")
```

### STEP 2: Displaying Basic Information

```python
# Display basic information
print(type(data))
print(data.head())
print(data.crs)
print(data.geometry.geom_type)
```

### STEP 3: Plot the ShapeFile

```python
data.plot()
```

### STEP 4: Plotting Geographic Data with Basemaps

```python
# Reproject the dataset to Web Mercator
data_web = data.to_crs(epsg=3857)
data_web['geometry'].head()
```

```python
# Create a map with an OpenStreetMap basemap
fig, ax = plt.subplots(figsize=(10, 10))

# Plot the shapefile data
data_web.plot(
    ax=ax,
    facecolor="none",
    edgecolor="red",
    linewidth=2
)

# Add OpenStreetMap basemap
ctx.add_basemap(ax, source=ctx.providers.OpenStreetMap.Mapnik)

ax.set_title("Study Area with OpenStreetMap Basemap")
ax.set_axis_off()

# Save the figure
plt.savefig(
    "./output_data/Ex1_clipping_polygon.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
